# **Creation Synthetic Data** 🤖

* In the Following Cells, we'll use a funtion to create our sinthetic data set. we'll emulate a **company that distributes** **osteosynthesis materials 🏥** across Americas, with three heeadquarters located in **US**🐱‍🏍, **Chile**🐱‍👓 and **Colombia**🐱‍💻.

```python
#=================================   Sinthecti Data  =============================================
def generate_synthetic_dataset():
    """
    This funtion Create a data set with synthetic data. Simulated a sales transations of Quirurgic Materials.
    Simulated Sales in USA, Chile and Colombia.
    Simulated Sales in period of 2024. For total sales of $5,000,000.00 USD
    """
    # Countries And Principal Citys  // You can Change countries and citys

    countries = {
        'Colombia': ['Bogotá', 'Medellín', 'Cali'],
        'Chile': ['Santiago', 'Valparaíso', 'Concepción'],
        'Estados Unidos': ['New York', 'Los Angeles', 'Chicago']
    }

    # Select 5 Headquarters
    branches = [
        ('Colombia', 'Bogotá'),
        ('Colombia', 'Medellín'),
        ('Chile', 'Santiago'),
        ('Estados Unidos', 'New York'),
        ('Estados Unidos', 'Los Angeles')
    ]

    # 10 items of Osteosyntesis Material
    items = [
        {'name': 'Titanium Bone Screw 4.5mm', 'stock_code': 'TBS045'},
        {'name': 'Locking Compression Plate 4.5mm', 'stock_code': 'LCP045'},
        {'name': 'Cortical Screw 3.5mm', 'stock_code': 'CS035'},
        {'name': 'DHS Screw 12.5mm', 'stock_code': 'DHSS125'},
        {'name': 'Intramedullary Nail 8mm', 'stock_code': 'IMN08'},
        {'name': 'Distal Femur Plate', 'stock_code': 'DFP'},
        {'name': 'Proximal Humerus Plate', 'stock_code': 'PHP'},
        {'name': 'Spinal Pedicle Screw', 'stock_code': 'SPS'},
        {'name': 'Bone Graft Substitute', 'stock_code': 'BGS'},
        {'name': 'Orthopedic Wire 1.25mm', 'stock_code': 'OW125'}
    ]

    # Clients 4 for Headquarter (Simulated 20 Hospitals or Clinics)
    clients = []
    for i, (country, city) in enumerate(branches):
        for j in range(4):
            client_id = f'CL_{i*4 + j + 1:03d}'
            clients.append({'client_id': client_id, 'branch': (country, city)})

    # Random Dates (2024)
    start_date = datetime(2024, 1, 1)
    end_date = datetime(2024, 12, 31)
    dates = [start_date + timedelta(days=random.randint(0, 364)) for _ in range(10000)]  # Pool de fechas

    # Random price for item (Between 10-1000 USD)
    item_prices = {item['stock_code']: np.random.uniform(10, 1000) for item in items}

    # Generate the trasations
    data = []
    sale_id = 1
    total_sales = 0
    target_sales = 5000000  # 5 Millions USD - you cand Level up or level down this Total  ;)

    while total_sales < target_sales and len(data) < 50000:  # IMPORTANT This is the Limit for total Transations.
        # Random Headquarter
        country, city = random.choice(branches)
        headquarters = f'{city}, {country}'

        # Random Client
        branch_clients = [c for c in clients if c['branch'] == (country, city)]
        client = random.choice(branch_clients)['client_id']

        # Random Item
        item = random.choice(items)
        name_item = item['name']
        stock_code = item['stock_code']
        price = item_prices[stock_code]

        # Ranndom Quantity (1-50 unids)
        unit = random.randint(1, 50)

        # Total Amout
        amount_prices = price * unit

        # Unit Cost (20-70% of price)
        cost_pct = random.uniform(0.2, 0.7)
        cost = price * cost_pct
        cost_amount = cost * unit

        #  Gross profit (amount_prices - cost_amount)
        operative_utility = amount_prices - cost_amount

        # Random Date
        date = random.choice(dates).strftime('%Y-%m-%d')

        data.append({
            'sale_id': sale_id,
            'name_item': name_item,
            'stock_code': stock_code,
            'headquarters': headquarters,
            'client_id': client,
            'date': date,
            'prices': round(price, 2),
            'unit': unit,
            'amount_prices': round(amount_prices, 2),
            'cost': round(cost, 2),
            'cost_amount': round(cost_amount, 2),
            'operative_utility': round(operative_utility, 2)
        })

        total_sales += amount_prices
        sale_id += 1

    # If the target is exceeded, adjust the last transation proportionally
    if total_sales > target_sales:
        overshoot = total_sales - target_sales
        last_row = data[-1]
        trim_factor = 1 - (overshoot / last_row['amount_prices'])
        last_row['unit'] *= trim_factor
        last_row['unit'] = max(1, round(last_row['unit']))  # Min 1 unit
        last_row['amount_prices'] = round(last_row['prices'] * last_row['unit'], 2)
        last_row['cost_amount'] = round(last_row['cost'] * last_row['unit'], 2)
        last_row['operative_utility'] = round(last_row['amount_prices'] - last_row['cost_amount'], 2)
        total_sales = target_sales

    # Create DataFrame
    df = pd.DataFrame(data)
    df['date'] = pd.to_datetime(df['date'])
    df = df.sort_values('date').reset_index(drop=True)  # short by date

    print(f"Dataset OK: {len(df)} Transations with TotalSales: ${total_sales:,.2f}")
    return df
```

# **9.** Time Series Analysis with Window Functions (Window Functions) 👀.
**Problem:**
 A business analyst asks you to calculate the 7-day moving average of operative utility for each headquarters of the company. This will help identify recent profitability trends.

### **Guide:**

* Ensure the date column is in the correct format (e.g., DateType).
* Define a window function that partitions the data by headquarters and orders by date.
* Apply a window range of 7 days (rowsBetween(-7, 0)) to include the current day and the previous 7 days.
* Calculate the average (avg) of the operative_utility column within each window and create a new column with the result.

## **Define the Window🐱‍👓**

>df_spark = df_spark.withColumn("date", F.to_date("date"))
>windowSpec = Window.partitionBy("headquarters").orderBy("date").rowsBetween(-7, 0)

## **Calculate 🐱‍🐉**

df_spark_with_ma = df_spark.withColumn(
    "rolling_avg_utility_7d",

## **Show Result💻**

> df_spark_with_ma.select("date", "headquarters", "operative_utility", "rolling_avg_utility_7d").show(truncate=False)


**Check the Ouput 😎**

# **10.** Profitability Analysis by Headquarters and Month👀🏛.
**Problem:**

A strategic stakeholder asks you to calculate the operative utility by headquarters and month. Additionally, they note it would be highly valuable to filter for the top 5 products beforehand.

### Guide:

* ***Filter by Year:*** Apply an initial filter to process only data from the year 2024.
* ***Month Extraction:*** Use Spark’s built-in F.month() function to extract the month from the date column. This will create a new column or be used directly in the groupBy.
* ***Aggregation by Granularity:*** Group your data by the headquarters, month, and name_item columns to calculate the sum of operative_utility at a monthly level for each product and headquarters.
* ***Window Function:*** Redefine the Window Spec to partition by the combination of headquarters and month. This ensures the product ranking resets and is calculated independently for each month and headquarters.
* ***Final Filter:*** Filter the resulting DataFrame to keep only records with a rank of 5 or lower.


## **Filtering for 2024🐱‍👓**

>df_2024 = df_spark.filter(F.year(F.col("date")) == 2024)

## **Groupby and calculate Total Profit by headquarter, Month and product🐱‍🐉**

>df_profitability = df_2024.groupBy("headquarters", F.month(F.col("date")).alias("month"), "name_item") \
                         .agg(F.sum("operative_utility").alias("total_utility"))


## **Defines and partitions the window by 'headquarters' and 'month'. For a monthly classification💻**

> windowSpec = Window.partitionBy("headquarters", "month").orderBy(F.desc("total_utility"))

## **Apply the sorting function📋**
>df_ranked = df_profitability.withColumn("rank", F.rank().over(windowSpec))

## **Filter to get the 5 most profitable products by location and month🧭**
>df_top_5_monthly = df_ranked.filter(F.col("rank") <= 5)

## **Show Result☕**
>df_top_5_monthly.show(truncate=False)*

## **Optional: save in parquet format💾**
>df_top_5_monthly.write.partitionBy("headquarters", "month").parquet("ruta/a/datos_optimizados_mensuales")

**Check print output**

```python
df_2024 = df_spark.filter(F.year(F.col("date")) == 2024)


df_profitability = df_2024.groupBy("headquarters", F.month(F.col("date")).alias("month"), "name_item") \
                         .agg(F.sum("operative_utility").alias("total_utility"))


windowSpec = Window.partitionBy("headquarters", "month").orderBy(F.desc("total_utility"))


df_ranked = df_profitability.withColumn("rank", F.rank().over(windowSpec))


df_top_5_monthly = df_ranked.filter(F.col("rank") <= 5)

df_top_5_monthly.show(truncate=False)

```
